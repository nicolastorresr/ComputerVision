# Understanding Biological Foundations of Vision

Desde el siglo XIX, científicos han estudiado cómo el ojo y el cerebro procesan la información visual. Hermann von Helmholtz propuso la "teoría de la visión", donde la percepción es una inferencia inconsciente y activa. Sus ideas influyeron en la visión computacional, además de desarrollar la teoría del color y el oftalmoscopio.

<div align="center">
    <img src="img/ophthalmolscope.png" alt="" width="30%" align="center" />
</div>

En 1943, Warren McCulloch y Walter Pitts, inspirados por el cerebro humano, crearon un modelo computacional de la actividad neuronal. Este modelo sentó las bases conceptuales de las futuras redes neuronales.

En el ámbito del hardware, el influyente artículo de Claude Shannon de 1948, "Una teoría matemática de la comunicación", mostró cómo medir la información, la transmisión de señales y los límites de compresión, algo crucial para el procesamiento de señales de visión artificial y el reconocimiento de patrones.

<div align="center">
    <img src="img/shannon.png" alt="" width="50%" align="center" />
</div>

Los hallazgos de Hubel y Wiesel, que les valieron el Premio Nobel en 1981, revelaron la organización jerárquica y modular de la corteza visual. Las neuronas en diferentes niveles procesan características visuales cada vez más complejas, lo que refleja el enfoque de muchos sistemas de visión artificial modernos. Su trabajo inspiró los primeros algoritmos de visión artificial que imitaban el procesamiento jerárquico observado en el cerebro.

<div align="center">
    <img src="img/Hubel-Wiesel.png" alt="" width="40%" align="center" />
</div>

Uno de los trabajos pioneros fue el sistema "Pandemonium" propuesto por Oliver Selfridge en 1959. Se trataba de un modelo de procesamiento paralelo jerárquico con "demonios" que procesaban los estímulos visuales en etapas, cada grupo se encargaba de un aspecto específico del reconocimiento:

- **Demonio de la imagen:** Captura la imagen inicial tal y como aparece en la retina.

- **Demonios de las características:** Se especializan en reconocer características visuales específicas como líneas o curvas. Cada demonio "grita" al detectar su característica.

- **Demonios cognitivos:** Escuchan los "gritos" de los demonios de las características y responden en función de la presencia de patrones que están entrenados para reconocer. La fuerza de su respuesta depende de la coincidencia del patrón detectado con su característica designada.

- **Demonio de la decisión:** Escucha las respuestas de los demonios cognitivos y elige la más fuerte, determinando de manera efectiva la percepción final.

# Deep neural networks

## A perceptron unit

Es el bloque básico de una red neuronal. Consiste en una capa de neuronas donde cada entrada tiene un peso asociado que se multiplica por la entrada y se suma todo. Si esa suma supera un umbral, la neurona se activa.

<div align="left">
    <img src="img/SLP.jpg" alt="" width="20%" align="left" />
</div>

## Weights

Son valores que se asignan a las conexiones entre las neuronas. Estos pesos determinan la importancia de la señal de entrada para la neurona. A través del entrenamiento, estos pesos se ajustan para que la red aprenda y haga predicciones precisas.

$$ s_i = \mathbf{w}_{i}^T\mathbf{x}_i + b_i  $$

Los $\mathbf{x}_i$ son las entradas de la $i$-ésima unidad, $\mathbf{w}_i$ y $b_i$ son sus pesos y sesgos que se pueden aprender, $s_i$ es la salida de la suma lineal ponderada.

## Regularization and normalization

La regularización es una técnica utilizada para prevenir el sobreajuste (overfitting) en el entrenamiento de la red. Ayuda a controlar la complejidad de la red al agregar términos a la función de pérdida que penalizan los pesos grandes.

### Dataset augmentation

Es una técnica para aumentar la cantidad de datos de entrenamiento mediante la aplicación de transformaciones pequeñas y aleatorias a las imágenes. Por ejemplo, rotar ligeramente una imagen, cambiar el brillo o recortarla de manera aleatoria. Esto enriquece el conjunto de datos y ayuda a mejorar la generalización y la capacidad de la red para reconocer patrones.

### Dropout

Es una técnica de regularización en redes neuronales que consiste en "apagar" aleatoriamente un cierto porcentaje de neuronas durante el entrenamiento. Esto fuerza a la red a aprender de manera más robusta, ya que no puede depender excesivamente de ninguna neurona en particular.

## Loss functions

Es una medida que indica qué tan bien está realizando la red sus predicciones en comparación con las respuestas reales. El objetivo es minimizar esta función durante el entrenamiento para mejorar la precisión de la red.

La función de pérdida de entropía cruzada (cross-entropy) es una métrica utilizada para medir qué tan bien una red neuronal de clasificación está realizando predicciones en comparación con las respuestas reales.

$$\text{Cross-Entropy} = -\frac{1}{N} \sum_{i=1}^{N} \left( y_i \cdot \log(p_i) + (1 - y_i) \cdot \log(1 - p_i) \right)$$

Esta métrica mide cuánta información se pierde en promedio al usar las predicciones de la red para representar las verdaderas etiquetas.


## Backpropagation

Es un algoritmo que se utiliza para entrenar una red neuronal. Consiste en propagar el error desde la salida hacia atrás a través de la red, ajustando los pesos de las conexiones en función de la contribución de cada peso al error.

## Convolutional neural networks

Es un tipo de red neuronal que utiliza capas convolucionales que aplican filtros para extraer características y reducir la cantidad de parámetros, lo que las hace eficientes en el procesamiento de imágenes.

### Pooling

Es una operación utilizada en las redes neuronales convolucionales para reducir la dimensionalidad de cada mapa de características, preservando la información más relevante. Por ejemplo, el max pooling selecciona el valor máximo en una región específica de los datos.

### Network architectures

#### VGG

Utiliza bloques convolucionales con filtros pequeños (3x3) y capas densas, utilizando convoluciones múltiples para aprender características. Se diferencia por su profundidad, con variantes que van desde VGG11 hasta VGG19, donde el número indica la cantidad de capas.

<div align="center">
    <img src="img/vgg.png" alt="" width="60%" align="left" />
    <img src="img/symbols.jpg" alt="" width="30%" align="center" />
</div>

#### Inception

Destaca por su módulo "Inception", que utiliza diferentes tamaños de filtros de convolución (1x1, 3x3, 5x5) en paralelo y luego los concatena para aprender características a diferentes escalas. Esto ayuda a reducir la cantidad de parámetros y mejorar la eficiencia computacional.

<div align="center">
    <img src="img/inception.jpg" alt="" width="45%" align="center" />
</div>

#### ResNet

Conocida por sus conexiones residuales. Introduce "skip connections" o conexiones directas que saltan capas, permitiendo que la red aprenda no solo la representación de las características, sino también las diferencias entre las capas. Esto ayuda a mitigar el problema de desvanecimiento de gradientes en redes muy profundas.
