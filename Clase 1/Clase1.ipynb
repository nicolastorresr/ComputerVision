# Línea de tiempo

Cuando nace la visión por computador se consideraba que el componente de percepción visual sería un aspecto clave para dotar a los robots de un comportamiento inteligente y que sería fácil de lograr. Sin embargo, resultó ser un problema más difícil de lo que se pensaba.

# 1960s - Fundamentos y optimismo inicial

En 1966, en el laboratorio de IA del Massachusetts Institute of Technology (MIT), Marvin Minsky — uno de los fundadores de la inteligencia artificial — le propuso a su estudiante Gerald Sussman un proyecto que supuestamente podría resolverse en un verano.

<div style="background-color:#d4edda; border-left:6px solid #28a745; padding:10px;">

**La idea era básicamente:**
    
Construir un sistema que pudiera “ver” — es decir, interpretar una escena visual capturada por una cámara.
</div>

No era simplemente clasificar dígitos.
El objetivo era algo mucho más ambicioso: que la máquina pudiera analizar una imagen y comprender la estructura del mundo físico que representaba (Qué objetos hay, Dónde están, Cómo están orientados, etc.).



<div style="background-color:#fff3cd; border-left:6px solid #ffc107; padding:10px;">

¿Por qué pensaban que era tan “simple” de lograr?
    
</div>

# Antes del Deep Learning

Durante años, la Visión por Computador se basaba en:

- Feature engineering manual (SIFT, HOG, SURF)
- Clasificadores tradicionales (SVM, Random Forest)


Limitaciones: Alta dependencia del diseño manual, baja capacidad de generalización y sensibilidad a iluminación, rotación, ruido.

# 2010-2017: ImageNet Large Scale Visual Recognition Challenge

Por primera vez, la comunidad tenía suficientes datos para entrenar modelos grandes: 14 millones de imágenes, 1000 categorías de objetos, anotaciones manuales, benchmark público.

# 2012: ImageNet Classification with Deep Convolutional Neural Networks

Alex Krizhevsky, Ilya Sutskever, Geoffrey Hinton

Redujo el error top-5 de ImageNet de 26% a 15%

1️⃣ dataset masivo (ImageNet)

2️⃣ Uso de GPU, ReLU, dropout, data augmentation

3️⃣ redes neuronales profundas (AlexNet)

# Arquitectura AlexNet

5 capas convolucionales; 3 capas fully connected; ~60 millones de parámetros; Uso de ReLU activations.

# ¿Por qué ReLU fue tan importante?

Las redes con ReLU crean regiones lineales por partes en el espacio de entrada.

La red aprende una función altamente no lineal.

# 2012: El punto de inflexión

- Explosión del Deep Learning en visión.
- Convolutional Neural Networks superan ampliamente métodos tradicionales.
- Se abandona progresivamente el feature engineering manual.
- La representación es aprendida automáticamente.
- Comienza la era de los grandes datasets.

# 2020s - Modelos fundacionales y visión multimodal

- Modelos cada vez más grandes y pre-entrenados con menos supervisión
- Vision Transformers (ViT) reemplazando o complementando las CNN
- Modelos multimodales que integran visión y lenguaje (CLIP, DALL-E, Stable Diffusion)
- Foundation models para visión (similar a GPT en NLP)
- Arquitecturas auto-supervisadas que requieren menos datos etiquetados
- Avances en neuro-rendering y Neural Radiance Fields (NeRF) para visión 3D
- Democratización con modelos open-source y herramientas accesibles
- Aplicaciones generalizadas en robótica, medicina, vehículos autónomos, etc.
- Enfoque en interpretabilidad, robustez y ética
